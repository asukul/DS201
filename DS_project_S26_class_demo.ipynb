```python
import pandas as pd
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/asukul/DS201/refs/heads/master/datasets/Automobile_price_data_clean2.csv")
```

```python
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the visual style
sns.set(style="whitegrid")

# Create a figure with multiple subplots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Distribution of Price
sns.histplot(df['price'], kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Distribution of Price')

# 2. Price by Body Style (Boxplot)
sns.boxplot(x='body-style', y='price', data=df, ax=axes[1])
axes[1].set_title('Price by Body Style')

# 3. Horsepower vs Price (Scatter plot)
sns.scatterplot(x='horsepower', y='price', hue='fuel-type', data=df, ax=axes[2])
axes[2].set_title('Horsepower vs Price')

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a pairplot for numeric columns, colored by fuel-type
sns.pairplot(df, hue='fuel-type', diag_kind='kde')
plt.show()
```
